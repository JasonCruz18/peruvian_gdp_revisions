# Tutorial 03: Data Cleaning

Learn how to clean and standardize GDP tables using 70+ cleaning functions.

```python
import sys
from pathlib import Path
import pandas as pd
sys.path.insert(0, str(Path.cwd().parent))

from peru_gdp_rtd.config import get_settings
from peru_gdp_rtd.cleaners import NewTableCleaner, OldTableCleaner
from peru_gdp_rtd.cleaners import clean_columns_values, normalize_text

settings = get_settings('../config/config.yaml')
```

## Step 1: Load Raw Table

```python
# Load a raw extracted table
from peru_gdp_rtd.processors import extract_table

# Example: Extract from sample PDF
sample_pdf = Path('../new_weekly_reports/2020/NS_01_enero_2020.pdf')
if sample_pdf.exists():
    tables = extract_table(sample_pdf, pages=[3])
    raw_df = tables[0]
    print(f'Raw table shape: {raw_df.shape}')
    display(raw_df.head())
```

## Step 2: Initialize Cleaner

```python
cleaner = NewTableCleaner(
    df=raw_df,
    wr_number=1234,
    table_num=1,
    config=settings,
)
print('Cleaner initialized')
```

## Step 3: Apply Cleaning Pipeline

```python
cleaned_df = cleaner.clean()
print(f'Cleaned table shape: {cleaned_df.shape}')
display(cleaned_df.head())
```
