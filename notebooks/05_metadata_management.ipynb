# Tutorial 05: Metadata and Base-Year Management

Track base-year changes and create benchmark datasets.

```python
import sys
from pathlib import Path
import pandas as pd
sys.path.insert(0, str(Path.cwd().parent))

from peru_gdp_rtd.config import get_settings
from peru_gdp_rtd.transformers import MetadataHandler

settings = get_settings('../config/config.yaml')
```

## Step 1: Initialize Metadata Handler

```python
handler = MetadataHandler(config=settings)
print('Metadata handler initialized')
```

## Step 2: Load RTD Datasets

```python
monthly_rtd, quarterly_rtd = handler.load_rtd_datasets()
print(f'Monthly RTD shape: {monthly_rtd.shape}')
print(f'Quarterly RTD shape: {quarterly_rtd.shape}')
```

## Step 3: Create Benchmark Datasets

```python
monthly_benchmark, quarterly_benchmark = handler.create_benchmark_datasets()
print(f'Monthly benchmark shape: {monthly_benchmark.shape}')
print(f'Quarterly benchmark shape: {quarterly_benchmark.shape}')

display(monthly_benchmark.head())
```

## Step 4: Apply Base-Year Adjustments

```python
by_adjusted_monthly = handler.apply_base_year_adjustments(monthly_rtd)
by_adjusted_quarterly = handler.apply_base_year_adjustments(quarterly_rtd)

print('Base-year adjusted datasets created')
```

## Key Concepts

- **Base-Year Changes**: Peru changed base year multiple times (1990, 1994, 2007)
- **Benchmark Dataset**: Vintages before base-year changes
- **Base-Year Adjusted**: Remove revisions due to base-year changes
- **Metadata**: Track timing and magnitude of structural changes
