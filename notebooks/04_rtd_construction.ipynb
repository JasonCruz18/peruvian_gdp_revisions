# Tutorial 04: Real-Time Dataset Construction

Build vintage-format datasets from cleaned tables.

```python
import sys
from pathlib import Path
import pandas as pd
sys.path.insert(0, str(Path.cwd().parent))

from peru_gdp_rtd.config import get_settings
from peru_gdp_rtd.transformers import VintagesPreparator
from peru_gdp_rtd.transformers import apply_base_year_sentinel

settings = get_settings('../config/config.yaml')
```

## Step 1: Load Cleaned Table

```python
# Assume we have a cleaned DataFrame from Tutorial 03
cleaned_df = pd.DataFrame()  # Replace with actual cleaned data
print(f'Cleaned data shape: {cleaned_df.shape}')
```

## Step 2: Prepare Vintages

```python
preparator = VintagesPreparator(
    df=cleaned_df,
    base_year=2007,
    vintage_name='2020_01',
    config=settings,
)

monthly_vintage, quarterly_vintage = preparator.prepare_vintages()
print(f'Monthly vintage shape: {monthly_vintage.shape}')
print(f'Quarterly vintage shape: {quarterly_vintage.shape}')
```

## Step 3: Apply Base-Year Sentinels

```python
monthly_with_sentinel = apply_base_year_sentinel(
    df=monthly_vintage,
    base_year=2007,
)

display(monthly_with_sentinel.head())
```

## Key Concepts

- **Vintage**: A snapshot of data at a specific release date
- **Base Year**: Reference year for index calculations
- **Sentinel**: Marker for base-year changes
- **RTD Format**: Vintages × Target periods matrix
